## Qu'est-ce que la TEI ?

### Lou Burnard et Marjorie Burghart, **Qu’est-ce que la Text Encoding Initiative ?**, 2015

> La TEI a été d’abord développée, il y a plus de trente ans, comme un projet de recherche dans le champ alors émergent du « Humanities computing ». L’idée originelle était de proposer un **ensemble de recommandations** sur la façon dont les chercheurs devraient créer des **ressources textuelles** « lisibles par ordinateur », qui soient **adaptées aux besoins de la recherche** – dans la mesure où un consensus existait sur le sujet –, mais qui soient également extensibles, puisque ces besoins changent et évoluent.


## Définition de TEI

TEI est un set de balises prédéfini et documenté dans les TEI guidelines qui permet de procéder à une description « scientifique » et « sémantique » d’un texte.

## Objectif : penser le texte

La TEI fournit un cadre pour penser la nature du texte : elle constitue une sorte d’encyclopédie de notions textuelles associées à des balises et des règles sur la façon dont elles peuvent être combinées.

## Historique

- 1987 Établissement de la Text Encoding Initiative
- 1990 TEI P1, dir. Michael Sperberg-McQueen et Lou Burnard
- 1992-93 TEI P2, expansion
- 1994 TEI P3, première version complète
- 2000 Naissance du TEI Consortium
- 2001-04 TEI P4, introduction du XML
- 2007 TEI P5, abandon du SGML

## Principes de Poughkeepsie (1987)

1. Provide a standard format for data interchange in humanities research
2. Suggest principles for the encoding of texts in the same format
3. Define a recommended syntax for the format and a metalanguage for the description of text-encoding schemes : describe the new format and representative existing schemes both in that metalanguage and in prose
4. Propose sets of coding conventions suited for various applications
5. Include a minimal set of conventions for encoding new texts in the format
6. The guidelines are to be drafted by committees on text documentation, text representation, text interpretation and analysis, metalanguage definition and description of existing and proposed schemes, coordinated by a steering committee of representatives of the principal sponsoring organizations
7. Compatibility with existing standards will be maintained as far as possible

## Les grands points forts

- Rencontre des communautés : encodage centré sur plusieurs niveaux de **contenus intellectuels** :
    - *Visuel* : Aspect physique (codicologie, manques, foliation, etc.) et aspects paléographiques (abréviations et formes développée, etc.)
    - *Sémantique* : Toponymes, anthroponymes, dates, balisage des types de discours, désambiguïsation des références textuelles…
    - *Analytique* : Correspondance des parties du texte à une grille d'analyse (taxonomie, thésaurus, etc.)

- **Conçue par et pour les chercheur·euses** : L'objectif est d'aider les chercheur·euses à créer des ressources textuelles adaptées aux besoins de la recherche

- **Interopérabilité et pérennité** : Indépendant de tout environnement logiciel, permet l'échange et la collaboration

## TEI n'est pas un schéma

TEI (All) n’est pas un schéma à proprement parler, mais plutôt un framework, utile à la conception de son propre schéma. Il est fortement déconseillé d’utiliser un schéma contenant l’intégralité de la TEI.


## Un modèle pour de nombreuses applications

**Modèle abstrait** : Ensemble des concepts et les relations entre ces
concepts.

**Implémentations** : Retranscription des contraintes formelles par le
biais d’un schéma de validation.

## Aspect modulaire

Les Guidelines de la TEI définissent plusieurs centaines de concepts différents, avec les instructions détaillées sur les éléments qui peuvent être utilisés pour les représenter. La plupart des documents TEI n’a besoin que d’une petite partie de ce qui est fourni.

Les projets d’encodage utilisent bien souvent un sous-ensemble très restreint de la TEI : une documentation restreignant l’ensemble utilisé peut être rédigée à l’aide d’outils comme [Roma](https://roma.tei-c.org/).

- **Tei Lite** = designed to meet 90% of the needs of 90% of the TEI user community

- **Tei All** = inventaire complet qui contient plus de 570 éléments et 200 attributs

# Organisation des guidelines

## Les modules

Les modules sont des ensembles cohérents de balises.

- Les modules obligatoires pour renseigner les métadonnées et couvrir les balises les plus courantes.
    - tei : définition des classes, macros et types de données
    - textstructure : éléments de base pour structurer un texte de type livre
    - core : éléments incontournables pour tout type de document
    - header : en-tête TEI (métadonnées du document)

- Les modules additionnels qui correspondent à l'encodage d'un type de texte (poésie, dictionnaires, etc.) ou d'un type de données et d'informations (apparat critique, élément graphique, etc.)

## Composition d'un document TEI

Tout document TEI contient *a minima* deux parties :

- un **en-tête** qui contient les métadonnées du document → `<teiHeader>`
- le **texte** → `<text>`

## Structure minimale

```xml
<TEI xmlns="http://www.tei-c.org/ns/1.0">
 <teiHeader>
    <fileDesc>
        <titleStmt>
            <title></title>
        </titleStmt>
        <publicationStmt>
            <p></p>
        </publicationStmt>
        <sourceDesc>
            <p></p>
        </sourceDesc>
    </fileDesc>
 </teiHeader>
 <text>
    <body>
        <p></p>
    </body>
 </text>
</TEI>
```

## Le cas des teiCorpus

```xml
<teiCorpus version="3.3.0" xmlns="http://www.tei-c.org/ns/1.0">
     <teiHeader>
     <!--[en-tête du corpus]-->
     </teiHeader>
     <TEI>
         <teiHeader>
         <!--[en-tête du premier texte]-->
         </teiHeader>
         <text>
         <!--[premier texte du corpus]-->
         </text>
     </TEI>
     <TEI>
         <teiHeader></teiHeader>
         <text></text>
     </TEI>
</teiCorpus>
```

## Multiplicité de l'objet texte

- Document source : document physique qui a été numérisé et sur lequel s'appuie l'encodage
- Document numérisé : Le document image issu de la numérisation du document source
- Document encodé : Le document numérique en lui-même au format XML TEI

## Convention de nommage

- `*Stmt` : Contenu structuré en sous-éléments

- `*Desc` : Description en prose, pouvant admettre des sous-éléments

- `*Decl` : Déclarations associant sous-éléments et description en prose

## [Structures versifiées et théâtre](https://www.tei-c.org/release/doc/tei-p5-exemplars/html/tei_lite_fr.doc.html#vedr)

### Le paratexte (`<front>`)

- `<set>` : décrit le décor, le lieu, le temps et l'ordre d'entrée en scène (attention : il ne s'agit pas d'une didascalie courante mais d'un résumé du cadre spatio-temporel).
- `<performance>` : indique les conditions ou l'historique d'une représentation (dans la préface/postface).
 - `<castList>` : la liste des personnages, structurée avec `<castGroup>` et `<castItem>` contenant `<role>`, `<roleDesc>` et éventuellement `<actor>`.

## [Structures versifiées et théâtre](https://www.tei-c.org/release/doc/tei-p5-exemplars/html/tei_lite_fr.doc.html#vedr)

###  Le corps de la pièce (`<body>`) :

- Les divisions majeures (`<div type="act">`, `<div type="scene">`)
- didascalies ou indications scéniques (`<stage>`)
    - Les répliques des personnages à l'intérieur d'un élément `<sp>` (pour *speech*)
    - L'attribut `who="..."` sur `<sp>` identifie le locuteur de façon normalisée (relié à l'identifiant @xml:id du rôle dans la distribution)
        - L'élément `<speaker>` qui contient le nom du personnage tel qu'il apparaît avant sa réplique.
        - Le texte de la réplique lui-même, généralement balisé en prose (`<p>`) ou en vers (`<l>` regroupés dans `<lg>`)

## Exemple simplifié d'encodage d'une réplique

```xml
<sp who="#alceste">
  <speaker>ALCESTE</speaker>
  <stage type="delivery">poussant un soupir</stage>
<p>Non, il n'est point de sort plus digne d'envie...</p>

```

### Structuration matérielle

| Balise  | Utilisation                                                                  |
|---------|------------------------------------------------------------------------------|
| `<pb/>` | Changement de page avec un attribut `@n` pour le numéro de page qui commence |
| `<cb/>` | Changement de colonne                                                        |
| `<lb/>` | Changement de ligne                                                          |

Ce sont des éléments de type milestone (balises auto-fermantes), qui désignent un point dans le flux textuel au même titre que `<milestone/>` élément borne permettant de délimiter une partie section de texte sans l'encadrer de manière un peu différente que l'élément `<anchor/>`.

## Les références internes

1. Poser un `@xml:id` **une fois** dans le document, sur l'élément cible
2. Référencer cet élément sur d'autres éléments du document grâce à des attributs de pointage (`@spanTo`, `@ref`, `@facs`, `@target`…) dont la valeur commence par `#`.

```xml
<p>
  Il partit au matin
  <addSpan spanTo="#fin-ajout" place="margin"/>
  et marcha longtemps sur la route
  <anchor xml:id="fin-ajout"/>
  avant d'arriver au village.
</p>
```

- **Diplomatique** : Attention à l'aspect visuel du texte et aux spécificités de graphies ([exemple : Le Didascalicon d’Hugues de Saint-Victor sur Thélème](http://theleme.enc.sorbonne.fr/dossiers/vue100.php))
- **Critique** : Mise en parallèle des version d'un texte pour en révéler les variantes ([exemple : The Online Froissart](https://www.dhi.ac.uk/onlinefroissart/collatey.jsp?del=5&pb3=P604_1r&pb2=P71_13r&pb5=T41_1r&pb4=Sto_1r&pb7=H25_1r&pb6=Tou_8r&pb9=Rou-1_t_11r&pb8=Rou-1_t_11r&div0=ms.f.transc.Bes-1&pb10=Rou-1_t_11r&div10=ms.f.transc.Rou-1&GlobalWord=1&panes=9&img10=&img4=&disp4=pb&img3=&disp5=pb&disp2=pb&img6=&disp3=pb&img5=&disp10=pb&img8=&disp8=pb&img7=&disp9=pb&disp6=pb&disp7=pb&div9=ms.f.transc.Rou-1&img9=&div8=ms.f.transc.Rou-1&div7=ms.f.transc.H25&div6=ms.f.transc.Tou&div5=ms.f.transc.T41&div4=ms.f.transc.Sto&div3=ms.f.transc.P604&div2=ms.f.transc.P71&div1=ms.f.transc.B88-1&img0=&disp0=pb&disp1=pb&pb1=B88-1_1r&img2=&pb0=Bes-1_1r&img1=))

## Focus sur l'édition paléographique

- rendre compte de l'état du texte tel qu'il apparaît dans le manuscrit
- proposer une version régularisée ou une édition critique (on va se concentrer sur la normalisée).

**Documentation de référence :** [TEI Guidelines, chapitre 11 « Representation of Primary Sources »](https://tei-c.org/release/doc/tei-p5-doc/en/html/PH.html)

## Relier la transcription à la page matérielle

```xml
<TEI xmlns="http://www.tei-c.org/ns/1.0">
  <teiHeader>…</teiHeader>

  <facsimile>
    <surface xml:id="page1">
      <graphic url="manuscrit_p1.jpg" width="2000px" height="3000px"/>
      <zone xml:id="l1" ulx="120" uly="200" lrx="1850" lry="260"/>
      <zone xml:id="l2" ulx="120" uly="270" lrx="1850" lry="330"/>
    </surface>
  </facsimile>

  <text>
    <body>
      <p>
        <pb facs="#page1"/>
        <lb facs="#l1" n="1"/>Au commencement était le verbe
        <lb facs="#l2" n="2"/>et le verbe était auprès de Dieu.
      </p>
    </body>
  </text>
</TEI>
```

## Régularisation des graphies

| Balise | Usage |
|---|---|
| `<orig>` / `<reg>` | graphie originale / graphie régularisée |
| `<abbr>` / `<expan>` | abréviation / développement |
| `<sic>` / `<corr>` | erreur présumée / correction |

```xml
<p>
  <!-- Variante graphique -->
  Il estoit <choice>
    <orig>grãd</orig>
    <reg>grand</reg>
  </choice> seigneur.

  <!-- Abréviation -->
  <choice>
    <abbr>Mr</abbr>
    <expan>Monsieur</expan>
  </choice> de La Fontaine

  <!-- Correction d'une faute -->
  Il <choice>
    <sic>mangea</sic>
    <corr>mangeait</corr>
  </choice> tranquillement.
</p>
```

## Entités personnalisées

```xml
<?xml version="1.0" encoding="UTF-8"?>
<!DOCTYPE texte [
  <!ENTITY etc "et cetera">
  <!ENTITY s-long "&#x017F;">
]>
<texte>
  <p>Des pommes, des poires, &etc;</p>
  <p>Le &s-long;iecle</p>
</texte>
```

## Caractères non standards

```xml
<teiHeader>
  <encodingDesc>
    <charDecl>
      <glyph xml:id="et-tironien">
        <desc>Note tironienne pour « et »</desc>
      </glyph>
    </charDecl>
  </encodingDesc>
</teiHeader>

<p>pain <g ref="#et-tironien">⁊</g> vin</p>
```

## Les lacunes

`<gap>` ou `<unclear>` passage ne pouvant pas être restitué pour des raisons matérielles

```xml
<gap reason="illegible" agent="ink-blot"/>
<unclear> passage en partie illisible

<unclear agent="ink-blot" cert="low" resp="#EC">bas</unclear>
```

`<supplied>` restitution d’un passage manquant

```xml
<supplied reason="hole" source="#autreEd">les</supplied>
```


### Ajouts et surplus

`<add>` pour le texte ajouté et `<del>` pour le texte raturé.

```xml
<add place="above">nous</add>
<del> texte supprimé</del>

<del rend="overstrike">claustra</del>
```

## La description des mains

`<@hand>` dans le corps du texte <body> `<@handDesc>` et ses sous-éléments dans le `<teiHeader>`

```xml
<handDesc>xml
  <handNote xml:id="c2Hand1" scope="major">à l’encre noire</handNote>
  <handNote xml:id="c2Hand2" scope="minor">au crayon</handNote>
</handDesc>
```

`<handNote>` est répétable et peut contenir une localisation avec `@locus`.

```xml
<p>Du texte <add hand="#c2Hand2">ajout</add> etc.</p>
```
