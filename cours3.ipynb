# Partie 1 : édition paléographique (suite), TEI pour la description les sources manuscrites

EAD pour **inventorier** (description des fonds en arborescence), TEI pour **décrire** un manuscrit.

## [`<msdDesc>`](https://www.tei-c.org/release/doc/tei-p5-doc/fr/html/ref-msDesc.html)

- `<msIdentifier>` Identification du manuscrit (cote, lieu de conservation, etc.). Seul élément obligatoire
- `<msContent>` Détails sur le contenu intellectuel du manuscrit
- `<physDesc>` Description de la matérialité du manuscrit
- `<history>` Renseignement sur la provenance et l’histoire de la transmission du manuscrit


## [`<msIdentifier>`](https://www.tei-c.org/release/doc/tei-p5-doc/fr/html/ref-msIdentifier.html)

- Contient les informations pour une identification non ambigüe du manuscrit
- Premier élément (et seul obligatoire) du msDesc

### Quelques éléments du `<msIdentifier>`

- `<idno>` Identifiant du manuscrit (cote généralement)
- `<repository>` Instance de conservation
- `<msName>` Titre du manuscrit
- `<altIdentifier>` Autre identifiant (ancienne cote ou id numérique)
- Le `<head>` (facultatif) l’intitulé du manuscrit (ne permet pas l'identification univoque)


## Décrire les contenus : `<msContents>`, `<msPart>`, `<msFrag>`

| Élément | Usage |
|---|---|
| `<msContents>` | décrit les textes d'un manuscrit (via `<msItem>`) |
| `<msPart>` | parties d'origine distincte réunies dans **un même volume** |
| `<msFrag>` | fragments **dispersés** d'un même manuscrit d'origine |

`<msPart>` et `<msFrag>` ne se combinent pas avec `<msContents>`.

## Le [`<msContents>`](https://www.tei-c.org/release/doc/tei-p5-doc/fr/html/ref-msContents.html)

- Soit une série de paragraphes `<p>`
- Soit du contenu structuré avec plusieurs `<msItem>`

### Principaux éléments du msContents

- `<summary>` Résumé global du contenu du manuscrit, possiblement structuré en table des matières indiquant avec locus les différentes parties
- `<msItem>` Permet de décrire une partie du manuscrit constituant une unité cohérente de contenu.
- `<titlePage>` Description du contenu de la page de titre

## `<summary>`

```xml
<msContents>
  <summary>
   <locus from=”1r” to=”2v”>
    Première partie du manuscrit
   </locus>
   <locus from=”3r” to=”18v”>
    Seconde partie du manuscrit
   </locus>
  </summary>
</msContents>
```

## msItem

```xml
<msItem>
  <locus from="xxxx" to="xxxx">
   Folios concernés par la section
 </locus>
 <incipit>
   Premiers mots de la section
 </incipit>
 <explicit>
   Premiers mots de la section
 </explicit>
</msItem>
```

## Description matérielle du document : le [`<physDesc>`](https://www.tei-c.org/release/doc/tei-p5-doc/fr/html/ref-physDesc.html)

Le physDesc contient la description physique partielle ou complète du manuscrit, éventuellement structurée en utilisant [des éléments spécialisés](https://tei-c.org/release/doc/tei-p5-doc/fr/html/ref-model.physDescPart.html).

### Principaux éléments du physDesc

- `<objectDesc>` Description globale de l’objet en cours de
description (dimension, type de support, etc.)
- `<handDesc>` Description des mains intervenants dans le manuscrit
- `<decoDesc>` Description des éléments décoratifs, notamment des enluminures
- `<bindingDesc>` Description de la reliure et éventuellement ses précédentes reliures

```xml
objectDesc
<objectDesc>
 <supportDesc>[…]</supportDesc>
 <layoutDesc>[…]</layoutDesc>
</objectDesc>
```

## Histoire et transmission du manuscrit : le `<history>`

Rassemble les éléments servant à donner un historique partiel ou complet du manuscrit, notamment à travers des informations sur son parcours depuis sa création.

### Principaux éléments du `<history>`

- `<origin>` Information sur le contexte et les circonstances de création du manuscrit
- `<provenance>` Information sur un épisode précis du manuscrit, entre sa création et son acquisition
- `<acquisition>` Information sur les circonstances de l'entrée du manuscrit dans l'institution qui le détient

## 👩🏻‍💻 Exercice

À l’aide de la [notice de l’IRHT](https://jonas.irht.cnrs.fr/manuscrit/45533) (Histoire du (des) volume(s)), renseignez le `<history>`

```xml
<history>
 <origin>[…]</origin>
   <provenance>[…]</provenance>
   <acquisition>[…]</acquisition>
</history>
```

## Structurer un `<biblStruct>`

| Élément | Usage |
|---|---|
| `<analytic>` | l'article ou le chapitre (facultatif) |
| `<monogr>` | le livre ou la revue qui le contient (obligatoire) |
| `<series>` | la collection (facultatif) |

`@level` sur `<title>` : `a` article, `m` livre, `j` revue, `s` collection.

```xml
<biblStruct>
  <analytic>
    <author><persName><surname>Cerquiglini</surname>, <forename>Bernard</forename></persName></author>
    <title level="a">Éloge de la variante</title>
  </analytic>
  <monogr>
    <title level="j">Langue française</title>
    <imprint>
      <biblScope unit="volume">40</biblScope>
      <biblScope unit="page" from="25" to="40"/>
      <date when="1983">1983</date>
    </imprint>
  </monogr>
</biblStruct>
```

**À retenir :** ordre fixe `analytic` → `monogr` → `series` ; `<imprint>` et `<biblScope>` vont dans `<monogr>`.

# Partie 2 : S'autonomiser avec la TEI

## Un exemple de workflow avec la TEI

- Modélisation des documents
- Spécification d’un sous-ensemble de la TEI (fichier ODD)
- Génération d’un schéma et d’une documentation
- Encodage des documents
- Révision du modèle (processus itératif)

## Les Guidelines

Les Guidelines guident les pratiques d’encodage, mais ne sont pas un standard qui contraint ce que l’on doit faire.

Les guidelines permettent de bénéficier de l’expérience accumulée en matière de stratégie d’encodage, de méthodologie.

## Comment lire les Guideline

- `Exemples` Pour mieux comprendre ce qu’il faut mettre dans chaque balise
- `Modules` Lire la documentation en fonction du contenu à encoder
- `Élément` Pour aller voir le modèle, le contenu et contenant d’une balise
- `Classes` Pour comprendre plus en détails comment est structurée la TEI

## Les modules

**Les modules sont un ensemble cohérent de balises regroupées pour répondre à une problématique précise.**

- Type de document : pour la poésie, le théâtre, les manuscrits, le discours oral, etc.
- Type de données : pour la définition d'un apparat critique, des formules mathématiques, etc.

## Chaque module

- est **documenté** par un chapitre des Guidelines
- définit un certain nombre de **composantes** (éléments, classes, macros)
- correspond à une **problématique particulière**
- peut être **inclus ou non** dans une personnalisation (à part les 4 obligatoires)

## Rappel des quatre modules obligatoires

- `tei` : définition des classes, macros et types de données
- `core` : éléments disponibles dans tous les documents
- `header` : en-tête contenant les métadonnées du document
- `textstrutcure` : éléments de base pour structurer le texte de type livre

## Les classes

Les classes sont définies dans un module mais peuvent porter sur des éléments d’un autre. Les éléments héritent des propriétés de la classe à laquelle ils appartiennent.

- `att.class` Éléments qui partagent les mêmes attributs
- `model.class` Éléments qui sont situés aux mêmes endroits dans le document

![classes](./media/guidelines-schema.png)

## `att.class`

Ensemble d’éléments qui admettent les mêmes attributs

| Classe | Description | Exemples |
| --- | --- | --- |
| `att.global` | Attributs pouvant être utilisé avec tout élément | `@xml:id`, `@n`, `@rend`, `@xml:lang`, etc. |
| `att.linking` | Attributs pour lier les éléments entre deux | `@corresp`, `@next`, `@prev`, `@sameAs`, etc. |
| `att.ascribed` | pour discours action pouvant être attribué à un individu spécifique | `@who` |


## Attributs à retenir

### Attributs globaux (toujours disponibles)

- `@n` fournit un numéro ou un label (pas unique)
- `@rend` pour préciser un certain rendu graphique de l'élément dans la source ou la sortie
- `xml:id` fournit un identifiant unique à l'élement qui porte l'attribut
- `xml:lang` indique la langue du contenu textuel de l’élément (code langue ISO)

## Attributs fréquemment employés

- `@type` pour typer un élément
- `@target` lien interne
- `@ref` référence
- `@key` référence canonique

## `model.class`

Se définit par un ensemble d'endroits où les membres de la classe sont autorisés (`…Part`) mais implique également une forme de regroupement sémantique (`…Like`)

| Classe | Description |
| --- | --- | 
| `model.divPart` | Éléments qui peuvent être contenu par une `div` |
| `model.phrase` | Éléments pouvant apparaître au niveau de la phrase |
| `model.nameLike` | Regroupement sémantique pour les divers types de noms |

## Les macros

Les macros définissent des raccourcis pour désigner lesvaleurs attendues au sein des éléments/attributs.

- types de contenus standardisés (par exemple [macro.paraContent](https://www.tei-c.org/release/doc/tei-p5-doc/en/html/ref-macro.paraContent.html)) ;
- valeurs autorisées (par exemple [data.count](https://www.tei-c.org/release/doc/tei-p5-doc/fr/html/ref-teidata.count.html)).

Elles sont utilisées dans les modèles de contenu pour faire référence à des séquences rencontrées dans des éléments.

## Le modèle de contenu

[`macro.limitedContent`](https://tei-c.org/release/doc/tei-p5-doc/en/html/ref-macro.limitedContent.html)

```xml
<content>
 <alternate minOccurs="0" maxOccurs="unbounded">
 <textNode/>
 <classRef key="model.limitedPhrase"/>
 <classRef key="model.inter"/>
 </alternate>
</content>
```
[`<desc>`](https://tei-c.org/release/doc/tei-p5-doc/en/html/ref-desc.html)

```xml
<content>
 <macroRef key="macro.limitedContent"/>
</content>
```

## Les espaces de nom

- si XML est la **langue**, l'espace de noms est le **lexique**
- propriété d'appartenance d'un élément à une **application XML** (un vocabulaire)
- chaque espace de noms est identifié par une **URI unique**

**Pourquoi ?** Un même document peut mêler plusieurs vocabulaires. Le _namespace_ a donc deux fonctions :

1. distinguer des éléments de même nom,mais de sens différents
2. permettre à la machine de traiter chaque vocabulaire correctement

## Implémentation : préfixes

- Un **préfixe** est attaché aux éléments et attributs : `prefix:localPart` (= *QName*)
- Il est associé à une URI par `@xmlns:prefix = "identifiant"`
- Déclaré sur la racine (usage dans tout le document) ou sur l'élément qui contient les éléments concernés
- Plusieurs déclarations possibles sur un même élément

## Exemple : la racine d'un fichier TEI

```xml
<TEI xmlns="http://www.tei-c.org/ns/1.0"
  xmlns:xs="http://www.w3.org/2001/XMLSchema"
  xmlns:m="http://www.w3.org/1998/Math/MathML"
  xmlns:sch="http://purl.oclc.org/dsdl/schematron"
  xmlns:xi="http://www.w3.org/2001/XInclude"
  xmlns:rng="http://relaxng.org/ns/structure/1.0">
```

Ici : 

- `xmlns="…"` : espace de noms **par défaut** (éléments non préfixés = TEI)
- 5 **préfixes** : usage ponctuel d'autres vocabulaires (`<m:math>`, `<sch:rule>`…)

## Espace de nom déclaré dans l'élément englobant

```xml
<svg xmlns="http://www.w3.org/2000/svg" width="12cm" height="10cm">
  <ellipse rx="110" ry="130"/>
  <rect x="4cm" y="1cm" width="3cm" height="6cm"/>
</svg>
```

- `svg`, `ellipse`, `rect` : tous dans l'espace de noms SVG, **pas besoin de préfixe**
- La déclaration s'applique à l'élément qui le déclare et à ses descendants

## Cas des attributs

- Le namespace par défaut ne s'applique pas aux attributs
- Aucune ambiguïté car un attribut dépend toujours de son élément porteur

Pour appartenir à un espace de nom différent de celui de son élément porteur, un attribut doit être **préfixé** :

```xml
<p xlink:type="simple" xlink:href="ellipses.html">
```
