Le langage XSLT

La standardisation de l’encodage induite par le respect des guidelines permet d’utiliser des [scénarios de transformation génériques](https://github.com/TEIC/Stylesheets).

Les TEI Stylesheets sont une bibliothèque de feuilles XSLT (3.0) et XSL-FO développée par Sebastian Rahtz et maintenue par le Consortium TEI.


Elles permettent de convertir des documents TEI vers de nombreux formats (XHTML, LaTeX, PDF, EPUB, DOCX, Markdown, etc.) et inversement !

Intérêts

gain de temps pour les transformations courantes
bonne couverture des usages éditoriaux standards
paramétrage riche sans devoir tout programmer
projet communautaire et open source (partage des « profiles »)
Limites

tous les éléments/attributs TEI ne sont pas traités
toutes les convertions complètes et/ou bidirectionnelles
une personnalisation XSLT devient vite nécessaire pour les projets éditoriaux avancés

## Les quatres instructions principales

`<xsl:copy/>`

`<xsl:copy-of/>`

`<xsl:value-of/>`

`<xsl:apply-templates/>`

### `<xsl:copy/>`

Copie de la balise matchée par le @match,
sans les namespaces, attributs, texte, etc.
• <xsl:copy/> peut contenir d’autres
règles, du texte, etc.
• Intérêt → copier un élément pour appliquer
des règles à ses enfants.
<xsl:template match="/TEI/text/body">
<xsl:copy>...</xsl:copy>
</xsl:template>

### `<xsl:copy-of>`

• Copie à l’identique de la balise matchée par
le @select et de ses nœuds enfants
(balises, attributs, textes).
• Élément vide (pas de règles internes).
• Impossible de modifier les éléments copiés !
• Intérêt → reproduire rapidement une partie
de l’arbre que l’on ne veut pas modifier.
<xsl:template match="/TEI/text/body">
<xsl:copy-of select="."/>
</xsl:template>

### `<xsl:value-of>`

• Renvoie uniquement la valeur textuelle de
la balise matchée par le @select.
• Élément vide (pas de règles internes).
• Intérêt → copier du texte sans les balises
(utile pour LATE
X).
<xsl:template match="/TEI/text/body">
<xsl:value-of select="."/>
</xsl:template>

• Quelle différence entre ces templates ?
<!-- 1 -->
<xsl:template match="//body/p">
<xsl:copy-of select="."/>
</xsl:template>
<!-- 2 -->
<xsl:template match="//body/p">
<xsl:copy>
<xsl:value-of select="."/>
</xsl:copy>
</xsl:template>

<xsl:apply-templates/> est une
instruction récursive : le processeur va
examiner les nœuds enfants de la balise
matchée par le @match dans l’ordre et
appliquer les règles qui leur sont associées

Sans elle, le processeur s’arrête à
l’emplacement désigné par le @match du
<xsl:template/> et ne traite pas les
éléments enfants.
• Exemple : la balise <TEI/> (cf. fichier).
<xsl:template match="/">
<TEI>
<xsl:apply-templates/>
</TEI>
</xsl:template>

• @select permet d’appliquer les règles
uniquement au nœud sélectionné.
• Ex. → inverser <teiHeader/> et <text/>
(cf. fichier) :
<xsl:template match="/">
<TEI>
<xsl:apply-templates select="//text"/>
<xsl:apply-templates select="//teiHeader"/>
</TEI>
</xsl:template>

Une première transformation
La transformation la plus courante est certainement celle qui permet de passer d’un fichier DOCX à un fichier TEI !

Ouvrez l’archive hugo.docx avec OxygenXML
Dans le navigateur d’archives sélectionnez le fichier word/document.xml
OxygenXML propose plusieurs scénarios de transformation préconfigurés accessibles depuis le menu

sélectionnez le scénario DOCX TEI P5 (transformation définie par les TEI Stylesheets)
appliquez le scénario
En ligne de commande on indiquera simplement dans un terminal :


docxtotei hugo.docx hugo.tei.xml

TEI P5 - XHTML
Nous l’avons vu, d’autres scénarios sont disponibles pour effectuer des transformations vers ou depuis la TEI. Tentons maintenant de produire une sortie HTML.

OxygenXML :

fichier à transformer : hugo.xml
scénario de transformation : TEI P5 XHTML
Ligne de commande :

teitohtml hugo.xml hugo.html

Paramétrer les sorties
Les feuilles de transformation disposent de paramètres prédéfinis que l’on peut changer afin de modifier la sortie.

Commençons par retirer la numérotation des chapitres.

OxygenXML :

fichier à transformer : hugo.xml
éditez le scénario précédent (OxygenXML proposera de le dupliquer)
dans Paramètres, modifiez numberHeadings pour désactiver la numérotation des chapitres
relancez la transformation et comparez avec la sortie précédente
Ligne de commande (utilisation de saxon)

saxon /tmp/tei/share/xml/tei/stylesheet/html/html.xsl -o:hugo.html -s:./hugo.xml numberHeadings=false
De nombreux autres paramètres sont accessibles afin d’adapter rapidement la sortie. Par exemple, pour les gestion des notes :

footnoteBackLink : 'true'
autoEndNotes : 'true'

Cascading Style Sheet
Essayons maintenant de rendre notre sortie HTML un peu plus séduisante avec un CSS personnalisé :

cssFile : 'chemin/vers/hugo.css'
Il est également possible d’indiquer une feuille de style pour l’impression avec le paramètre cssPrintFile.

Inclure/importer une XSLT
Outre l’utilisation des paramètres, les TEI Stylesheets peuvent être appelées à l’intérieur d’une autre feuille XSLT.

<xsl:import/> : les règles du module dans lequel on importe ont la prééminence sur les règles du module importé
<xsl:include/> : pas de prééminence (erreur ou alerte)
Exercice

Créez une nouvelle feuille de transformation XSLT
 inclure la feuille de transformations html.xsl :
Chemin OxygenXML : ${frameworks}/tei/xml/tei/stylesheet/html/html.xsl
en ligne : https://www.tei-c.org/release/xml/tei/stylesheet/html/html.xsl
 ajoutez les différents paramètres utilisés précédemment avec <xsl:param/>

<xsl:param name="numberHeadings" select="'false'"/>
<xsl:param name="autoEndNotes" select="'true'"/>
<xsl:param name="footnoteBackLink" select="'true'"/>

<xsl:param name="documentationLanguage" select="'fr'"/>
<xsl:param name="lang" select="'fr'"/>
<xsl:param name="doclang" select="'fr'"/>

<xsl:param name="cssFile" select="'/chemin/vers/hugo.css'"/>

Enfin, appliquez la transformation sur le fichier hugo.xml.
