# Cours 5 : Le langage XSLT

## L'environnement XML

🌳 XSL → spécifications pour écrire des feuilles de style (stylesheet) ;

🌳 Une feuille de style → un doc XML avec l'extension `.xsl` qui contient des instructions ou règles écrites dans le langage de programmation XSLT.

🌳 XSLT → un langage conçu pour transformer des documents XML en d'autres documents selon les spécifications XSL.

🌳 Transformation opérée par un processeur XSLT (Saxon par exemple) qui lit le document source, le transforme en un autre arbre selon les règles de la feuille de style, et sérialise l'arbre de sortie dans le format souhaité.

## Les feuilles de transformation

- Élément racine : `<xsl:stylesheet>`
- Une première instruction pour déclarer ce qu'on veut en sortie
- Des instructions ou règles (templates)

```
<?xml version="1.0" encoding="UTF-8"?>
<xsl:stylesheet xmlns:xsl="http://www.w3.org/1999/XSL/Transform"
    xmlns:tei="http://www.tei-c.org/ns/1.0"
    xpath-default-namespace="http://www.tei-c.org/ns/1.0"
    xmlns="http://www.tei-c.org/ns/1.0"
    exclude-result-prefixes="tei"
    version="2.0">

    <xsl:output method="xml" indent="yes"/>

    <xsl:template match="//teiHeader">
        <xsl:copy>
            <xsl:apply-templates select="//fileDesc"/>
            <xsl:copy-of select="//encodingDesc"/>
            <xsl:copy-of select="//profileDesc"/>
            <xsl:copy-of select="//revisionDesc"/>
            <xsl:copy-of select="//fileDesc/following-sibling::*"/>
        </xsl:copy>
    </xsl:template>

</xsl:stylesheet>
```

👩🏻‍💻 Dans Oxygen → ouvrir un nouveau doc « XSLT Stylesheet » observer les attributs de l'élément racine. Ouvrir aussi [l'encodage du journal de Pierre Le Fèvre](https://github.com/gromettoclara/HNU6.3052/blob/main/exs-cours5/journal-Lefevre.xml).

## L'élément racine, attributs et espaces de nom

Espaces de nom (namespace) et attributs présents de base dans Oxygen :

- `@xmlns:xsl` : espace de nom XSL
- `@xmlns:xs` : XML Schema pour typer les variables et paramètres (`xs:string` `xs:integer`) ??
- `@xmlns:math` : fonctions mathématiques (XPath3)

- `@exclude-result-prefixes` : liste des préfixes qui ne seront pas copiés dans l'output
- `@version` : version de XSL (1, 2 ou 3)

## Comment adapter cet élément racine pour une transformation vers TEI

Déclarer l'espace de nom TEI pour que la feuille lise l'entrée et écrive la sortie proprement.

- `xmlns:xs → xmlns:tei` (et mise à jour de `exclude-result-prefixes`)
- ajout de `xpath-default-namespace` (lecture de l'entrée)
- ajout de `xmlns` (écriture de la sortie)

La méthode vaut pour tout standard XML (EAD, etc.) et pour HTML.

## <xsl:template> → définir une règle

- Possède un `@match` avec pour valeur un chemin XPath qui donne l'emplacement du nœud sur lequel sera appliquée la règle.
- Une feuille de transformation peut être composé d'une seule règle ou d'une succession de règle.
- Important → cohérence des XPath.

### `<xsl:copy>`

Copie de la balise matchée par le `@match`, sans les namespaces, attributs, texte, etc.

- `<xsl:copy>` peut contenir d'autres règles, du texte, etc.
- Intérêt → copier un élément pour appliquer des règles à ses enfants.

```
<xsl:template match="/TEI/text/body">
    <xsl:copy>...</xsl:copy>
</xsl:template>
```

### `<xsl:copy-of>`

- Copie à l'identique de la balise matchée par le `@select` et de ses nœuds enfants
(balises, attributs, textes).
- Élément vide (pas de règles internes).
- Impossible de modifier les éléments copiés !
- Intérêt → reproduire rapidement une partie de l'arbre que l'on ne veut pas modifier.

```
<xsl:template match="/TEI/text/body">
    <xsl:copy-of select="."/>
</xsl:template>
```

### `<xsl:value-of>`

- Renvoie uniquement la valeur textuelle de la balise matchée par le `@select`
- Élément vide (pas de règles internes)
- Intérêt → copier du texte sans les balises (utile pour LaTeX par exemple)

```
<xsl:template match="/TEI/text/body">
    <xsl:value-of select="."/>
</xsl:template>
```

## 👩🏻‍💻 Appliquer une règle sur les paragraphes

- Créer un `<xsl:template>`
- Indiquer le chemin XPath vers les `<p>` dans l'`@match`
- Copier à l'identique les nœuds avec leurs enfants

Ce qui donne :

```
<xsl:template match="/TEI/text/body/div/p">
    <xsl:copy-of select="."/>
</xsl:template>
ou
<xsl:template match="//body/p">
    <xsl:copy-of select="."/>
</xsl:template>
```

## 👩🏻‍💻 Quelle différence entre ces templates ?

```
<!-- 1 -->
<xsl:template match="//body/p">
    <xsl:copy-of select="."/>
</xsl:template>

<!-- 2 -->
<xsl:template match="//body/p">
    <xsl:copy>
        <xsl:value-of select="."/>
    </xsl:copy>
</xsl:template>
```

## Condition : `<xsl:if>` ou `<xsl:choose>`

- `<xsl:if>` ne possède pas de *else*, mais on peut les multiplier
- `<xsl:choose>` a pour enfant un ou plusieurs `<xsl:when>` (req.) et un seul `<xsl:otherwise>` (opt.) qui correspond au *else*
- Les règles de `<xsl:if>` et `<xsl:when>` sont appliquées si le contenu du `@test` est validé
- L'`@test` contient un XPath

## Boucles : `<xsl:for-each>`

- `<xsl:for-each>` a un `@select`
- Les règles contenues sont appliquées sur chaque nœud désigné par le `@select`
- ⚠️ Le point de départ des chemins XPath à l'intérieur du `<xsl:for-each>` est la balise sélectionnée dans le `@select`
- Peut intégrer des conditions

👩🏻‍💻 *Exemple : lister les titres des divisions, et signaler celles qui ont plus de 3 paragraphes.*

```
   <xsl:output method="text" indent="yes"/>

   <xsl:template match="/">
         <xsl:for-each select="TEI/text/body/div">
            <xsl:value-of select="head"/>
            <xsl:if test="count(p) > 3">
               <hi> (division longue)</hi>

            </xsl:if>
            <xsl:text>&#10;</xsl:text>
         </xsl:for-each>
   </xsl:template>
```

## Des balises peuvent être écrites directement dans un `<xsl:template>`

```
<xsl:template match="//text/body/div[1]">
 <TEI>
  <body>
     <div>
        <head type="chap">
           <xsl:value-of select="./head"/>
           <xsl:copy-of select="./p"/>
        </head>
     </div>
  </body>
 </TEI>
</xsl:template>
```

## `<xsl:element>` et `<xsl:attribute>`

Les balises XML ou HTML peuvent être écrites dans des `<xsl:element name="tag">` et les `<xsl:attribute name="@att">`

```
<xsl:template match="//body">
  <xsl:element name="TEI">
  <xsl:element name="div">
     <xsl:element name="head">
        <xsl:attribute name="type">chap</xsl:attribute>
        <xsl:value-of select="./div[1]/head"/>
     </xsl:element>
     <xsl:copy-of select="./div[1]/p"/>
  </xsl:element>
  </xsl:element>
</xsl:template>
```

## Un template to kill them all : ne fonctionne pas 🫨

```
<xsl:template match="/">

  <!-- ꩜ Boucle n°1 : on parcourt les <p> nous-mêmes -->
  <xsl:for-each select="p">
    <p>

      <!-- ꩜ Boucle n°2 : on parcourt TOUS les enfants de <p> un par un (texte, <hi>, etc.) -->
      <xsl:for-each select="node()">
        <!-- 😵‍💫 On « demande » à chaque nœud ce qu'il est -->
        <xsl:choose>
          <xsl:when test="self::hi">
            <!-- instruction de traitement -->
          </xsl:when>
          <xsl:otherwise>
            <!-- par exemple un nœud test : instructions de traitement -->
          </xsl:otherwise>
        </xsl:choose>
      </xsl:for-each>
    </p>
  </xsl:for-each>
</xsl:template>
```

## Un template to kill them all ne fonctionne pas → `<xsl:apply-templates/>` 🙏

`<xsl:apply-templates/>` est une instruction récursive : le processeur va examiner les nœuds enfants de la balise matchée par le `@match` dans l'ordre et appliquer les règles qui leur sont associées.

👩🏻‍💻 Essayer :

```
<xsl:template match="/">
    <TEI>
        <xsl:apply-templates/>
    </TEI>
</xsl:template>
```

Par défaut, si elle ne trouve aucune règle associée, l'instruction copie le texte.

👩🏻‍💻 créer deux règles récursives pour traiter les `<p>` et les `<hi>`

```
<xsl:template match="/">
    <TEI>
        <xsl:apply-templates/>
    </TEI>
</xsl:template>

<xsl:template match="p">
  <p><xsl:apply-templates/></p>
</xsl:template>

<xsl:template match="hi">
  <hi><xsl:apply-templates/></hi>
</xsl:template>
```

*Que se passe-t-il si les éditeur.ices rajoutent des notes dans les paragraphes et/ou dans les éléments en emphase ?*

Il suffit de rajouter un template !

## L'`@select` de `<xsl:apply-templates/>`

`@select` permet d'appliquer les règles uniquement au nœud sélectionné.

👩🏻‍💻 *Exemple → je veux exactement le même document mais inverser `<teiHeader>` et `<text>`*

```
<xsl:template match="/">
  <TEI>
     <xsl:apply-templates select="//text"/>
     <xsl:apply-templates select="//teiHeader"/>
  </TEI>
</xsl:template>
```

# XSLT4lazy

La standardisation de l'encodage induite par le respect des guidelines permet d'utiliser des [scénarios de transformation génériques](https://github.com/TEIC/Stylesheets).

Les TEI Stylesheets sont une bibliothèque de feuilles XSLT (3.0) et XSL-FO développée par Sebastian Rahtz et maintenue par le Consortium TEI.

Elles permettent de convertir des documents TEI vers de nombreux formats (XHTML, LaTeX, PDF, EPUB, DOCX, Markdown, etc.) et inversement !

## Intérêts

- gain de temps pour les transformations courantes
- bonne couverture des usages éditoriaux standards
- paramétrage riche sans devoir tout programmer
- projet communautaire et open source (partage des « profiles »)

## Limites

- tous les éléments/attributs TEI ne sont pas traités
- une personnalisation XSLT devient vite nécessaire pour les projets éditoriaux avancés

Les Stylesheets sont distribuées sous plusieurs formes :

- [Repo Github](https://github.com/teic/stylesheets), pour une installation locale et une utilisation en ligne de commande
- Paquet Debian
- Framework installé par défaut et accessible dans OxygenXML

## Inclure/importer une XSLT

Outre l'utilisation des paramètres, les TEI Stylesheets peuvent être appelées à l'intérieur d'une autre feuille XSLT.

- `<xsl:import>` : les règles du module dans lequel on importe ont la prééminence sur les règles du module importé
- `<xsl:include>` : pas de prééminence (erreur ou alerte)
- La feuille de transformation est appelée dans l'`@href`

👩🏻‍💻 Exercice :

- Créez une nouvelle feuille de transformation XSLT
- inclure la feuille de transformations `html.xsl` :
- Chemin OxygenXML : `/tei/xml/tei/stylesheet/html/html.xsl`
- en ligne : https://www.tei-c.org/release/xml/tei/stylesheet/html/html.xsl
- ajoutez les différents paramètres utilisés précédemment avec `<xsl:param>`

```
<xsl:param name="numberHeadings" select="'false'"/>
<xsl:param name="autoEndNotes" select="'true'"/>
<xsl:param name="footnoteBackLink" select="'true'"/>

<xsl:param name="documentationLanguage" select="'fr'"/>
<xsl:param name="lang" select="'fr'"/>
<xsl:param name="doclang" select="'fr'"/>

<xsl:param name="cssFile" select="'/chemin/vers/hugo.css'"/>
```

Enfin, appliquez la transformation sur le fichier `hugo.xml`.
