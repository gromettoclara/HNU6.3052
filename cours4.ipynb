# Cours 3 : XDM, XPath, Validation du document

# Partie 1 : XDM → [structure définie par le W3C](http://www.w3.org/TR/xpath-datamodel/)

- **XDM** : la structure de données, un **arbre** de nœuds (document, éléments, attributs, texte, commentaires…). C'est ce que manipulent XPath, XSLT et XQuery.
- **Fichier texte `.xml`** : une **sérialisation**, c'est-à-dire une façon d'écrire cet arbre sous forme de texte, pour le stocker ou l'échanger.

## Les 7 types de nœuds de l'arbre XDM

1. `Document Node`
2. `Element Node`
3. `Attribute Node`
4. `Text Node`
5. `Namespace Node`
6. `Processing Instruction Node`
7. `Comment Node`

![](public/media/XMD.png)

```xml
<bibliotheque>
    <livre id="01">
        <titre>Le Hobbit</titre>
    </livre>
</bibliotheque>
```

Représentation d'une hiérarchie :

```
• [Nœud Document] (Racine de l'arbre)
└── [Nœud Élément] bibliotheque
    └── [Nœud Élément] livre
        ├── [Nœud Attribut] id (valeur: "01")
        └── [Nœud Élément] titre
        └── [Nœud Texte] "Le Hobbit"
```

## Caractéristiques de l'arbre XDM

- Chaque nœud possède une identité unique
- Le document a un ordre (`Document Order`)

# Partie 2 : XPath

## Structure en arbre 🌳

Le langage Xpath fournit une syntaxe pour sélectionner un nœud ou un ensemble de nœuds au sein d’un document XML.

## Comment pratiquer ?

Oxygen permet de surligner les nœuds sélectionnés par un chemin entré dans une zone de saisie `Fenêtre` > `Afficher la vue` > `Constructeur Xpath / XQuery`.

## Les principes du langage XPath

- Un standard du W3C
- Un **langage de requêtes** qui permet de parcourir un arbre XML
- XPath est conçu comme un **langage intégré** (il n'est pas autonome)
- Une syntaxe de type chemin : `node/childnode`
    - Une expression de chemin correspond à une **séquence d’étapes** séparées par l’opérateur `/` .
    - Sans indication particulière la relation se fait d’un élément parent vers un élément enfant.
    - Chaque étape devient le nœud courant (nœud de contexte) pour l’étape suivante.

## Chemin absolu vs. chemin relatif

Un chemin de localisation **absolu** commence au nœud document : **_il commence par une barre oblique `/`_**

Les chemins de localisation **relatifs** commencent à un nœud de l’arborescence. Ils sont utilisés notamment au sein d’un autre chemin Xpath.

## Désignation d’une localisation

Une expression XPath définit un chemin de localisation, constitué de pas de localisation ("étape du chemin") séparés par le caractère `/`.

## Deux opérateurs bien pratiques…

**`*` Remplace n’importe quel nœud**

`/TEI/teiHeader/*` : Tout ce qui est contenu dans le `teiHeader`

`/TEI/teiHeader/*/div` : Toute `div` contenue dans un sous-élément du `teiHeader`

`/TEI/teiHeader//div` : fonctionne aussi…

**`|` Pour combiner deux ensembles de nœuds**

 `/TEI/teiHeader | /TEI/text/body` : Le `teiHeader` et le `body`

L’ordre des chemins doit suivre l’ordre d’apparition dans le document. Par exemple, `/TEI/text|/TEI/teiHeader` ne sélectionne que le text.

Qu'est-ce qui est sélectionné par `/TEI/(teiHeader/fileDesc|text/body)/*/p`

Les paragraphes contenus dans le `fileDesc` et le `body`.

## Axe

Indique la direction dans laquelle se déplacer dans l'arbre XML, relativement au nœud courant ou depuis la racine. Quand l'axe n'est pas précisé, il s'agit implicitement de l'axe des enfants (`child::`).

`/TEI/*/p/attribute::n`

`/TEI/*/p/parent::*`

Qu'est-ce que j'attrape ?

### Combinaison

`/TEI/*/lg/(preceding-sibling::*|following-sibling::*)`

Qu'est-ce que j'attrape ?

### Syntaxe axe attribute

`/TEI/*/p/attribute::n`

La valeur de l’attribut `n` des éléments `p` contenus dans le document

### Syntaxe axe parent

`/TEI/*/p/parent::*`

Les éléments qui contiennent directement un `p`

### Combinaison

`/TEI/*/lg/(preceding-sibling::*|following-sibling::*)`

Les éléments au même niveau que les `lg` du document

## Prédicats

Les prédicats sont utilisés pour filtrer les nœuds sélectionnés par l'axe et le test de nœud. Ils sont écrits entre crochets. Si le prédicat est évalué à vrai, tous les nœuds correspondants sont sélectionnés.


### Noeud qui contient…

`//title[.="Sonnet d’automne"]`

Les éléments title qui ont pour contenu "Sonnet d’automne"

Cette syntaxe abrégée est l’équivalent de : `[text()="Sonnet d’automne"]`

### Noeud qui a un enfant qui…

`//titleStmt[title="Sonnet d’automne"]`

Les éléments `titleStmt` dont l’enfant `title` a pour contenu "Sonnet d’automne"

`//titleStmt[title="Sonnet d’automne"]/author`

Les éléments `author` contenus dans les éléments `titleStmt` dont les enfants `title` ont pour contenu "Sonnet d’automne"

### Nœud qui a un descendant qui…

`//div[.//persName="Cendrillon"]`

Les éléments `div` qui contiennent un élément `persName` ayant pour
contenu "Cendrillon"

### Test d’existence

`//titleStmt[title]`

Les éléments `titleStmt` qui contiennent un élément `title`

### Élément qui a pour attribut…

`lg[@type]`

Les éléments `lg` qui ont un attribut `type`

`lg[@type="quatrain"]`

Les éléments `lg` qui ont un attribut `type` ayant pour valeur "quatrain"

### Prédicat numérique

`lg[@type="quatrain"][2]`

Le deuxième élément `lg` qui a un attribut `type` ayant pour valeur "quatrain"

Cette syntaxe est une abréviation de `lg[position()=2]`.

Contrairement aux autres langages de programmation, l’énumération XPath commence avec 1

Les tests formulés dans les prédicats sont définis à l'aide d'opérateurs et de fonctions.

Les opérateurs et les fonctions permettent de combiner et comparer, tester, transformer ou exploiter les données à tester. Ils peuvent aussi d'appliquer à des ensembles de nœuds.

## Les opérateurs

| Opérateurs | Usage |
| --- | ---|
| `+ - * div mod` | Pour effectuer des opérations arithmétiques |
| `= !=` | Pour tester l’égalité ou l’inégalité |
| `< > <= >=` | Strictement plus grand / plus petit que (&lt; et &gt; en XSLT) / Inférieur / supérieur ou égal |
| `and or` | Pour accoler plusieurs expressions booléennes |

`l[@n > 2]` ?

`lg[@type != "tercet" and @n < 2]` ?

`l[@n > 2]` Les éléments `l` dont la valeur de l'attribut `n` est supérieur à 2

`lg[@type != "tercet" and @n < 2]` Les éléments `lg` dont la valeur de l’attribut `n` est strictement inférieure à 2 et dont l’attribut type n’est pas "tercet"

Équivalent à `lg[@type != "tercet"][@n < 2]`

## Quelques exemples de fonctions

_Sélectionner le dernier vers du poème_

`(//l)[count(//l)]` ⚠️ parenthèses pour regrouper tous les éléments `l` ; sans elles l'expression sélectionne parmi les frères et sœurs

`count()` compte le nombre de l dans le document ou parmi les frères et sœurs

_Sélectionner tous les `lg` sauf le premier_

`//lg[position() != 1]`


## Quelques exemples de fonctions

`//l[last()]`

`(//l)[last()]`

`last()` sélectionne le dernier nœud dans le document ou parmi les enfants de chaque parent qui contient `l`

`//persName[contains(., "duc")]`

# Partie 3 : Technologies pour la validation et customisation de la TEI

La vie d'un document XML est assurée par sa famille de standards dont font partie les langages de schéma.

- `DTD` pour Document Type Definition
- `Relax NG`
- `XML Schema`

\+ Schematron

`Un schéma` est document définissant un ensemble de règles concernant la structure et le contenu de fichiers XML pour permettre leur validation.

## DTD

```xml
<!ELEMENT texte (chapitre+)>
<!ELEMENT chapitre (titre?, paragraphe+)>
<!ELEMENT titre (#PCDATA)>
<!ATTLIST chapitre n CDATA #REQUIRED>
```

- Langage historique
- Définit la structure d'un document, les éléments et les attributs qui y sont autorisés.
- Pas de typage précis du contenu des éléments, pas de gestion des espaces de nom
- Un format qui n'est pas du XML, mais peut être intégré dans le fichier

## XML Schema

```xml
<xs:element name="texte" maxOccurs="unbounded">
    <xs:complexType>
        <xs:sequence>
            <xs:element name="titre" type="xs:string"/>
            <xs:element name="chap" type="xs:decimal" minOccurs="0"/>
        </xs:sequence>
    </xs:complexType>
</xs:element>
```

- Recommandation du W3C
- Permet de préciser le contenu autorisé des éléments (typage, nombre) ainsi que des séquences d'éléments
- Les fichiers XML schema sont eux-aussi des éléments XML

## Relax NG

```xml
<element name="edition">
    <zeroOrMore>
        <element name="texte">
        <element name="titre">
            <text/>
        </element>
        <element name="paragraphe">
            <text/>
        </element>
        </element>
    </zeroOrMore>
</element>
```

- Simplification de XML Schema
- Standard ISO
- Supporte les namespaces
- Compatibilité avec XML Schema pour le typage des données

## Schematron

```xml
<s:rule context="tei:body"><!-- où s'applique la règle -->
   <s:assert test="exists(tei:div[@type = 'entries'])"><!-- test en XPath -->
      body must contain a div of type "entries".<!--message si le test échoue -->
   </s:assert>
</s:rule>
```

- Langage intégré à Relax NG
- Permet de définir des contraintes complexes
- Tout ce qui n'est pas spécifié est considéré comme valide.

## Un schéma est une combinaison des trois

- `DTD` pour la définition des entités
- `XML Schema` ou `RNG` pour les règles de validation
- Règles schematron

## Le langage ODD

- Format XML TEI qui comporte à la fois un schéma de validation et une documentation de ce même schéma.

- Le langage ODD a été développé en même temps que la TEI afin de pouvoir documenter et formaliser les recommandations

- Une spécification ODD est un simple document XML TEI qui utilise le module `tagdocs`.

- Il est possible de la transformer en fichier `RelaxNG` (`.rng`) à l’aide d’une feuille de transformation XSLT ou du logiciel `Roma`.

- Il est aussi possible de sortir une documentation en HTML.

## Le Consortium met à disposition un certain nombre de customisations

`teilite` TEI allégée

`tei_corpus` TEI pour les corpus linguistiques

`tei_ms` TEI pour la description des manuscrits

`tei_drama` TEI pour les corpus dramatiques

`tei_allPlus` Toute la TEI avec des additions

## En quoi consiste la customisation

- **Supprimer** des éléments, des attributs ou des classes entières

- **Ajouter** des éléments au sein d'une classe, ou bien des valeurs d'attributs admis par une classe ou un élément

- **Modifier le modèle de contenu** d'un élément, par exemple les types de contenu et les séquences d'éléments qu'il admet

- **Renommer** des éléments et/ou leur usage par rapport aux _Guidelines_

## Types de modification à la TEI : _clean_ vs. _unclean_

- _clean_ : Le schéma est un sous-ensemble de la TEI All (il est dit _TEI conformant_)

- _Unclean_ : Le schéma considère comme valides des documents que la TEI All considère invalides

## Que signifie _TEI conformant_ ?

- `Fichier bien formé` : l'encodage constitue un fichier XML valide

- `Validation sur TEI all` : l'encodage peut être validé avec un schéma TEI `all`

- Respect des `principes généraux` : l'encodage respecte le modèle abstrait de la TEI

- `Documentation` : les choix d'encodage sont documentés

## I. ODD by example : pas à pas

### Les étapes sont les suivantes

1) Vérifier que le framework TEI P5 est actif
2) Créer une ODD
3) Créer un RNG avec transformation XSL
4) Associer le fichier RNG au document TEI

## I. ODD by example : pas à pas

### 1. Vérifier que le framework TEI P5 est actif

1. Ouvrez les Préférences d'Oxygen (menu `Options > Préférences`).
2. Dans la colonne de gauche, sélectionnez `Association de type de document`.
3. Dans la liste de droite, faites défiler jusqu'à la lettre T : vous devez y voir la ligne `TEI P5`.
4. Vérifiez que la case située à côté de TEI P5 est bien cochée (ce qui signifie qu'elle est activée).

## I. ODD by example : pas à pas

### 2. Mise en place du scénario 👩🏻‍💻

- Ouvrez [`Sonnet_Automne_TEI.xml`](https://github.com/gromettoclara/HNU6.3052/blob/main/exs-cours4/Sonnet_Automne_TEI.xml)
- `Document > Transformation > Configurer` (Ctrl + Maj + C)
- `Nouveau > XML transformation with XSLT`
- Cochez "Options globales" pour rendre le scénario applicable à n’importe quel document
- Nommez le scénario "ODD by example"
- Renseigner le chemin XSL
`${frameworks}/tei/xml/tei/stylesheet/tools/oddbyexample.xsl`
- Sélectionnez le processeur `Saxon-PE > 9.xX`
- ⚙️ `Options avancées > Template (-it)` > Remplir le champ avec la valeur "main"
- `Paramètres > Nouveau > Nom "corpus" et Valeur "${cfdu}"` (i.e. le répertoire courant) : si déjà défini, double-cliquez sur le paramètre pour en changer la valeur
- Onglet Sortie > Définir le nom de fichier `${cfd}/ODD_${cfn}.xml` pour la future ODD (qui s’enregistrera dans le même dossier que le XML transformé)
- Appliquer le scénario au fichier
