# Valider la structure d'un document suite

## XPath

### Structure en arbre : maîtrise de la hiérarchie XML

Le langage Xpath fournit une syntaxe pour sélectionner un nœud ou un ensemble de nœuds au sein d’un document XML.

### Comment pratiquer ?

Oxygen permet de surligner les nœuds sélectionnés par un chemin avec la barre **Xpath** + **Entrée**.

mettre screenshot

### Les principes du langage XPath

- Un standard du W3C
- Un **langage de requêtes** qui permet de parcourir un arbre XML
- XPath est conçu comme un **langage intégré** (il n'est pas autonome)
- Une syntaxe de type chemin : `node/childnode`
    - Une expression de chemin correspond à une **séquence d’étapes** séparées par l’opérateur `/` .
    - Sans indication particulière la relation se fait d’un élément parent vers un élément enfant.
    - Chaque étape devient le nœud courant (nœud de contexte) pour l’étape suivante.

### Chemin absolu vs. chemin relatif

Un chemin de localisation **absolu** commence au nœud document : **_il commence par une barre oblique `/`_**

Les chemins de localisation **relatifs** commencent à un nœud de l’arborescence. Ils sont utilisés notamment au sein d’un autre chemin Xpath.

Désignation d’une localisation

Une expression XPath définit un chemin de localisation, constitué de pas de
localisation (“étape du chemin”) séparés par le caractère / .
Les pas sont définis par :

Axe
Sens dans lequel l’arbre
doit être parcouru
parcouru (parent::,
descendant::)

Test de nœud
Nom d’un nœud ou
type de nœud situé
dans la direction de l’axe
dans l’arborescence

Prédicat(s)
Indiqué entre [...]
pour préciser une
caractéristique du noeud
(sorte de filtre)

syntaxe de localisation

```
[Prédicat(s)](axe::nœud[predicat1][predicat2]…)
```

### Types de noeuds courants

| `elementName`  | élément | Désigne un élément (balises et contenu |
| `@attributName` |  attribut  | Désigne l'attribut de l'élément XML  |
| `text()` |  texte | Désigne le contenu textuel **directement** contenu dans un élément (sans la balise) |
| `comment()` | Commentaire | Désigne le contenu d’un commentaire (pas son balisage <!--...-->) |

### Exemples

`/TEI/teiHeader/titleStmt/title` L'élément `title` et son contenu `<title type=”play”>Le Misanthrope</title>`

`/TEI/teiHeader/titleStmt/title/text()` La chaîne de caractère contenu dans l’élément `title` « Le Misanthrope »

`/TEI/teiHeader/titleStmt/title/@type` La valeur de l’attribut type de l’élément `title` « play »

Syntaxe axe attribute
/TEI/*/p/attribute::n

La valeur de l’attribut n des éléments p contenus dans le document

Syntaxe axe parent

## Les abréviations d’axe

| vide | child:: | Axe par défaut |
| @ | attribute:: | Sélectionne la valeur d’un attribut |
| // | /descendant::node()/ | Sélectionne parmi les descendants |
| .. | parent::node()|  Sélectionne le nœud parent |
| . | self::node() | Sélectionne le nœud courant |

`//p` ? tous les éléments p peu importe où dans l'arboresence

`//p/@n` ? la valeur des attributs n des éléments p

`//div//p` ? Tous les éléments p descendants d'un élément div dans le document

`.//p` ? Tous les éléments p descendant du nœud courant

`//p/..` ? Les éléments contenant directement un élément p

`//p/parent::div` ? Les div contenant directement un p

`//lg//text()` ? Le texte contenu dans les descendants de tous les éléments lg

Prédicats
Les prédicats sont utilisés pour filtrer les nœuds
sélectionnés par l'axe et le test de nœud. Les prédicats
sont écrits entre crochets. Si le prédicat est évalué à
vrai, tous les nœuds correspondants sont sélectionnés.

Noeud qui contient...
//title[.=”Sonnet d’automne”]

Les éléments title qui ont pour contenu “Sonnet d’automne”
Cette syntaxe abrégée est l’équivalent de :
[text()=”Sonnet d’automne”]

Les opérateurs

| `+ - * div mod` | Pour effectuer des opérations arithmétiques |

## Le langage ODD

Format XML TEI qui comporte à la fois un
schéma de validation et une documentation
de ce même schéma
Le langage ODD a été développé en même
temps que la TEI afin de pouvoir documenter et
formaliser les recommandations
Une spécification ODD est un simple document
XML TEI qui utilise le module tagdocs

Une ODD est un Document XML qu’il est possible de transformer
en fichier RelaxNG (.rng) à l’aide d’une feuille de
transformation XSLT ou du logiciel Roma
