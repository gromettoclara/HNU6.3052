## Les formats textuels

| | Texte brut | Texte balisé | Texte formaté |
|---|---|---|---|
| **Exemple** | `.txt` | `.md`, `.tex`, `.xml`, `.html` | `.odt`, `.doc`, `.pages` |
| **+** | Interopérable | Sémantisme, mise en forme basique, interopérabilité | Mise en forme personnalisée, interface graphique |
| **−** | Information minimale, pas de structure | Besoin d'être interprété / transformé | Besoin d'un logiciel, fichier complexe |

## Un document structuré

> Un document peut être décrit comme une collection d’objets comportant des objets de plus haut niveau composés d’objets plus primitifs. Les relations entre ces objets représentent les relations logiques entre les composants du document. Par exemple [...] un livre est divisé en chapitres, chaque chapitre en sections, sous-sections, paragraphes, etc. Une telle organisation documentaire est appelée représentation de document structuré. Jacques André, Richard Furuta, Vincent Quint, *Structured documents*, Cambridge University Press, 1989.

# XML

- Un _métalangage_ ;
- Pas de jeu de balise prédéfini ;
- Un modèle de contenu arborescent ;
- Format de données simple, documenté ;
- Conçu pour la **description** des documents textuels ;
- Pas de réelle sémantique, mais permet le développement de vocabulaires descriptifs de balisages interopérables spécifiques à certains domaines.

## Standard internationnal

Depuis 1998, XML est un langage libre et documenté. C’est un standard respectant les recommandations du W3C, il facilite :

- la lisibilité par les machines ou par l’œil humain ;
- l’échange de données ;
- la migration vers d’autres plateformes, d’autres logiciels, d’autres formats.

## Historique

- 1970' : SGML, Standard Generalized Markup Language ;
- 1989-1992 : HTML, HyperText Markup Language ;
- 1998 : XML, eXtensible Markup Language.

## Principes de conception

- XML ne sert pas à afficher les données mais à les décrire. Seul, il ne fait rien. Appliqué à la représentation des textes, il permet de décrire notamment leur structure

- Le nom des balises n’est pas prédéfini : on peut librement créer son propre vocabulaire

- On peut utiliser une "grammaire" de balises (un schéma), pour définir des contraintes ou s’assurer de la consistance de l’encodage

- XML est auto-descriptif et lisible par l’homme. Un fichier XML n’est rien que du texte, on peut facilement prendre connaissance d’un corpus dans n’importe quel éditeur textuel

## Principes syntaxiques

- **Les Balises** : Les données textuelles sont repérées de part et d’autres par des "étiquettes";

- **Imbrication** : Les éléments sont contenus les uns dans les autres et ne se chevauchent pas ;

- **Héritage** : Les éléments enfants héritent des éléments parents.

## Les langages XML

- **TEI** : encodage des documents patrimoniaux ;

- **EAD** : Desciption des fonds d'archive ;

- **XHTML** : Structure des pages web ;

- **SVG** : Documents vectoriels.

# De quoi est composé un document XML ?

- **Le prologue**

- **Les éléments** : Les balises ainsi que leur contenu

- **Les attributs** : Les “caractéristiques” d’une balise

- **Les entités** : Appel à un caractère ou une séquence prédéfinie

- **Les commentaires** : Pour extraire des passages du XML

```
<?xml version="1.0" encoding="UTF-8"?>
<root>
 <element><!-- comment --></element>
 <empty/>
 <el attribute="value">&amp;</el>
</root>
```

## Le prologue

Il indique la version de la recommandation XML qui sert de base à l'écriture du fichier (il n'y a eu à cette date qu'une seule version, la version 1.0), ainsi que l'encodage de caractères utilisé → `<?xml version="1.0" encoding="UTF-8"?>`

## Les éléments

```
<element>texte</element>
<elementVide/>
```

- **Un arbre d'éléments** : Qui permettent de décomposer le texte en unités d’information
- **Dénomination** : Un élément doit être entouré de deux balises du même nom (sensible à la casse)
- **Clotûre** : Un élément ouvert doit être fermé
- **Imbrication** : Un élément peut contenir du texte, et/ou d'autres élements, mais **il ne peut pas en chevaucher un autre**
- **Racine** : Il existe un et un seul élément englobant tous les autres, encodé en premier, juste après le prologue : c’est l’**élément racine**.

## Les attributs

```
<miseEnValeur position="centrePage">
 texte
</miseEnValeur>
```

- **À quoi ils servent ?** : Ils sont utilisés pour préciser le contenu de l'élément, lui attribuer une caractéristique
- **Où les placer ?** : Placé les uns à la suite des autres dans la balise ouvrante
- **Contenu** : Un attribut doit nécessairement avoir une valeur, même si cette valeur est nulle `attribut=""`
- **Répétabilité** : Un même attribut ne peut pas être réutilisé dans la même balise

## Règles d'écriture des identificateurs génériques et des attributs

La chaîne de caractères formant le nom **peut** contenir :

- les caractères alphanumériques (lettres de a à z, et de A à Z, chiffres de 0 à 9, caractères non latins),
- le tiret bas ou _underscore_ `_`,
- le trait d’union `-`
- et le point `.`

La chaîne de caractère formant le nom **ne peut pas** :

- contenir tout autre signe de ponctuation,
- contenir des espaces,
- commencer par un nombre, ni par un point, ni par un trait d’union,
- commencer par xml (quelle que soit la casse utilisée)

Attention, XML est sensible à la casse !

## Les entités

```xml
&entité;
```

Entités internes : entités caractères prédéfinies pour saisir certains caractères que l’on doit obligatoirement coder ( <, >, &, ", ' : caractères codés au moyen des entités lt, gt, amp, quot, apos)

Entités externes : XML propose des mécanismes pour donner un nom à des expressions ou phrases souvent répétées ou pour taper des caractères difficiles à composer au clavier.
- **Caractère** : Elles sont utilisées à la place d’un caractère interdit en XML comme le &
- **Remplacement** : Mise en place d'une séquence de code contenue dans une DTD


## Les commentaires

```
<!-- texteCommentaire -->
```

Servent à… commenter, utiles pour documenter ce qu’on fait ou consigner des notes. Ne sont pas destinés aux programmes informatiques, qui les ignorent par défaut.

- Balise ouvrante = balise fermante

- 1 seul élément racine

- Respect de l'imbrication

- 1 attribut du même nom par balise

## OHCO, Ordered Hierarchy of Content Objects

Une certaine modélisation du texte :

Avec ce type de balisage, le texte est envisagé comme une hiérarchie ordonnée d’objets de contenus (OHCO, _Ordered Hierarchy of Content Objects_, en anglais).

- structure hiérarchique
- relations linéaires

## Bien formé ?

```
<paragraphe>du texte</paragraphe>

<paragraphe><article>du</article><nom>texte</nom></paragraphe>

<paragraphe><article>du <nom></article>texte</nom></paragraphe>

<paragraphe><article>du <nom>texte</nom></article></paragraphe>

<paragraphe type="texte">du texte</paragraphe>

<paragraphe type=texte>du texte</paragraphe>
    
<paragraphe type="texte">du texte<paragraphe/>
    
<paragraphe type="texte">du texte<nom>nom</paragraphe>
    
<paragraphe type="texte">du texte</Paragraphe>
    
<segment type="texte" type="nombre">du texte</paragraphe>
```
